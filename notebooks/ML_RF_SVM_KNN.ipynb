```python
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.multiclass import OneVsRestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import roc_auc_score
```

```python
# 1. Load your extracted embeddings
print("Loading data...")
data = np.load('../dataset/training_data_embeddings_all.npz')
X = data['X']
Y = data['Y']

# 2. Split into Train and Validation sets (to check your score locally)
X_train, X_val, Y_train, Y_val = train_test_split(X, Y, test_size=0.2, random_state=42)
print(f"Training on {X_train.shape[0]} samples, Validating on {X_val.shape[0]} samples.")
```

## Random Forest

```python
print("Training Random Forest...")
model = RandomForestClassifier(n_estimators=200, class_weight='balanced', n_jobs=-1, random_state=42)
model.fit(X_train, Y_train)

print("Predicting on validation set...")
# RF multi-label predict_proba returns a list of arrays. We must stack them.
prob_list = model.predict_proba(X_val)
val_preds = np.array([prob[:, 1] for prob in prob_list]).T

# BirdCLEF is evaluated on ROC AUC
roc_auc = roc_auc_score(Y_val, val_preds, average='macro')
print(f"Local Validation ROC AUC: {roc_auc:.4f}")

# Save the trained model to a file to upload to Kaggle
joblib.dump(model, 'random_forest_model.pkl')
print("Model saved to random_forest_model.pkl")
```
